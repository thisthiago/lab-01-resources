# O que é um RDD?

RDD (Resilient Distributed Dataset) é a estrutura de dados fundamental do Spark. Ele representa uma coleção imutável (que não pode ser alterada) de objetos que podem ser processados em paralelo em um cluster.

Transformações vs Ações:

 - Transformações (map, filter, etc.) são lazy (não executam imediatamente)

 - Ações (collect, count, etc.) forçam a execução


1. Persistência: Você pode armazenar RDDs em memória com rdd.persist() para reutilização
2. Particionamento: RDDs são divididos em partições para processamento paralelo
3. Imutabilidade: RDDs não podem ser alterados, apenas transformados em novos RDDs

```scala

# O contexto do Spark (sc) já está disponível no Zeppelin
print("SparkContext disponível:", sc)
```

```scala
dados = [1, 2, 3, 4, 5]
rdd = sc.parallelize(dados)

# Coletar e mostrar os dados
print(rdd.collect())  # [1, 2, 3, 4, 5]
```

```scala
# Dobrar todos os valores (map)
rdd_dobro = rdd.map(lambda x: x * 2)
print(rdd_dobro.collect())  # [2, 4, 6, 8, 10]

# Filtrar apenas números pares (filter)
rdd_pares = rdd.filter(lambda x: x % 2 == 0)
print(rdd_pares.collect())  # [2, 4]
```

```scala
rdd = sc.parallelize([1, 2, 3, 4, 5])

# Contar elementos
print("Count:", rdd.count())

# Soma
print("Sum:", rdd.sum())

# Primeiro elemento
print("First:", rdd.first())

# Pegar os 3 primeiros
print("Take 3:", rdd.take(3))
```

```scala
texto = ["olá mundo", "olá spark", "mundo big data"]
rdd_texto = sc.parallelize(texto)

# Passo 1: Dividir as frases em palavras (flatMap)
palavras = rdd_texto.flatMap(lambda frase: frase.split(" "))

# Passo 2: Mapear cada palavra para um par (palavra, 1)
pares = palavras.map(lambda palavra: (palavra, 1))

# Passo 3: Reduzir somando as contagens
contagem = pares.reduceByKey(lambda a, b: a + b)

# Resultado final
print(contagem.collect())
# Resultado: [('olá', 2), ('mundo', 2), ('spark', 1), ('big', 1), ('data', 1)]
```
